## Importando bibliotecas

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Importando o Dataset

```python
df = pd.read_excel('C:/Users/Usuario-Pc/Desktop/default_of_credit_card_clients__courseware_version_1_21_19.xls')
```

## Exercício 1 — Definindo as listas de características financeiras

```python
bill_feats = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3',
              'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']

pay_feats  = ['PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3',
              'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']
```

## Exercício 2 — Resumo estatístico das faturas (BILL_AMT)

```python
df[bill_feats].describe()
```

## Exercício 3 — Distribuição dos valores de fatura (histogramas)

```python
df[bill_feats].hist(bins=20)
plt.tight_layout()
plt.show()
```

## Exercício 4 — Resumo estatístico dos pagamentos (PAY_AMT)

```python
df[pay_feats].describe()
```

## Exercício 5 — Distribuição dos valores de pagamento (histogramas)

```python
df[pay_feats].hist(bins=20)
plt.tight_layout()
plt.show()
```

## Exercício 6 — Identificando pagamentos iguais a zero

```python
mask_zero = df[pay_feats] == 0
mask_zero.sum()
```

## Exercício 7 — Transformação logarítmica dos pagamentos não nulos

```python
pay_nonzero = df[pay_feats][~mask_zero.any(axis=1)]
pay_log = pay_nonzero.apply(np.log10)
pay_log.hist(bins=20)
plt.tight_layout()
plt.show()
```
